# INTRODUCCION

La industria de los videojuegos es uno de los mercados de entretenimiento más dinámicos y competitivos del mundo. Para las empresas dedicadas a la comercialización de videojuegos, comprender los factores que influyen en el éxito de un título es fundamental para optimizar la toma de decisiones comerciales, identificar oportunidades de crecimiento y diseñar estrategias de marketing más efectivas.

En este proyecto se analizarán datos históricos de ventas de videojuegos recopilados hasta el año 2016 por la tienda online Ice, que comercializa videojuegos a nivel global. El conjunto de datos incluye información relevante sobre las ventas de juegos, las plataformas en las que están disponibles, los géneros, las reseñas de usuarios y críticos especializados, así como las clasificaciones de edad asignadas por la Entertainment Software Rating Board (ESRB).

El objetivo principal del análisis es identificar patrones y características asociadas al éxito comercial de los videojuegos. A partir de estos hallazgos, se buscará obtener información valiosa que permita detectar proyectos prometedores y apoyar la planificación de campañas publicitarias para el año 2017.

Para alcanzar este objetivo, se llevará a cabo un proceso estructurado que incluye la exploración y preparación de los datos, el análisis de tendencias de ventas por plataforma y género, la evaluación de la influencia de las calificaciones de usuarios y críticos en el desempeño comercial de los juegos, y la construcción de perfiles de usuario para diferentes regiones geográficas. Finalmente, se realizarán pruebas de hipótesis estadísticas para determinar si existen diferencias significativas entre las calificaciones promedio de determinadas plataformas y géneros.

El análisis permitirá transformar los datos históricos en información estratégica, proporcionando una base sólida para la toma de decisiones comerciales y la identificación de oportunidades dentro del mercado global de videojuegos.

## Inicialización

````python
# Cargar todas las librerías
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Carga los archivos de datos en DataFrame

games = pd.read_csv('/datasets/games.csv')
````

## Preparar los datos

Antes de realizar el análisis exploratorio y las pruebas estadísticas, fue necesario llevar a cabo un proceso de limpieza y preparación de los datos con el fin de garantizar la calidad y consistencia de la información utilizada.

En primer lugar, se estandarizaron los nombres de las columnas convirtiéndolos a minúsculas. Esta práctica facilita la manipulación de los datos y evita errores relacionados con diferencias entre mayúsculas y minúsculas al momento de realizar consultas y transformaciones.

````python
# Convertir todos los nombres de columnas a minúsculas
games.columns = games.columns.str.lower()

# Verificar el resultado
print(games.columns)
````

````python
#Verificar información de la base de datos
print("Games:", games.info())
print("")
print("Games:", games.shape)
print("")
print('\nValores nulos:')
print("Games:", games.isna().sum())
print("")
print('\nDuplicados:')
print("Games:", games.duplicated().sum())
````

Con base en la información de games.info(), las columnas que requieren conversión de tipo de dato son principalmente year_of_release y user_score.

````python
# Convertir año de lanzamiento a entero (permitiendo valores nulos)
games['year_of_release'] = games['year_of_release'].astype('Int64')
````

La columna year_of_release se encontraba almacenada como tipo float64. Esto suele ocurrir cuando existen valores ausentes, ya que Pandas utiliza el tipo flotante para representar números con valores nulos. Dado que el año de lanzamiento representa un valor entero, la columna fue convertida al tipo Int64, que permite trabajar con números enteros conservando los valores ausentes.

````python
# Reemplazar TBD por NaN y convertir user_score a numérico

# Convertir a texto y pasar a minúsculas
games['user_score'] = games['user_score'].astype(str).str.lower()

# Reemplazar tbd por NaN
games['user_score'] = games['user_score'].replace('tbd', np.nan)

# Convertir a numérico
games['user_score'] = pd.to_numeric( games['user_score'],errors='coerce')

#verificar 
games['user_score'].dtype
games['user_score'].isna().sum()
````

La columna user_score estaba almacenada como tipo object debido a la presencia de valores de texto, específicamente la abreviatura "TBD" y sus variantes posibles Tbd, TBd, TbD ("To Be Determined"). Estos registros indican que la puntuación aún no había sido determinada al momento de recopilar los datos. Como no representan una calificación válida, fueron reemplazados por valores ausentes (NaN) y posteriormente la columna fue convertida a un tipo numérico (float64) para facilitar cálculos estadísticos y análisis de correlación.

Las demás columnas presentaban tipos de datos adecuados para la información que contenían. Las variables de ventas regionales (na_sales, eu_sales, jp_sales y other_sales) permanecieron como valores numéricos de tipo flotante, mientras que las variables categóricas como platform, genre y rating conservaron el tipo object.

Valores ausentes en name y genre

Las columnas name y genre presentan únicamente dos valores ausentes cada una. Debido a que representan una proporción mínima del conjunto de datos y además corresponden a información fundamental para identificar y clasificar los videojuegos, estos registros pueden eliminarse sin afectar significativamente el análisis.

Valores ausentes en year_of_release

La columna year_of_release contiene 269 valores ausentes. Aunque representan una pequeña proporción del total de registros, el año de lanzamiento es una variable clave para los análisis temporales. Debido a que no es posible inferir de manera confiable el año correcto para cada videojuego, estos valores se conservarán como ausentes y se excluirán únicamente de aquellos análisis que requieran información temporal.

Una posible explicación para estos valores faltantes es que algunos títulos antiguos o de plataformas menos populares no contaban con información completa al momento de recopilar la base de datos.

Valores ausentes en critic_score

La columna critic_score presenta 8578 valores ausentes, más de la mitad del conjunto de datos. Esto sugiere que una gran cantidad de videojuegos no recibió evaluaciones por parte de críticos especializados o que dichas puntuaciones no estaban disponibles en las fuentes consultadas.

Debido a la elevada cantidad de datos faltantes, reemplazarlos mediante promedios podría introducir sesgos importantes en el análisis. Por esta razón, los valores ausentes se conservarán como NaN.

Valores ausentes en user_score

La columna user_score contiene 6701 valores ausentes. Parte de estos valores provienen de registros originalmente etiquetados como "TBD" ("To Be Determined"), que fueron convertidos a valores nulos durante la preparación de los datos.

La ausencia de puntuaciones de usuarios puede deberse a que algunos juegos no recibieron suficientes reseñas o a que las evaluaciones aún no estaban disponibles al momento de la recopilación de los datos. Al igual que en el caso de critic_score, los valores ausentes se mantendrán sin imputar para evitar distorsionar los resultados.

Valores ausentes en rating

La columna rating presenta 6766 valores ausentes. Una posible explicación es que ciertos videojuegos no fueron clasificados por la ESRB, especialmente títulos antiguos, juegos exclusivos de determinadas regiones o productos dirigidos a mercados distintos del norteamericano.

Dado que la clasificación ESRB constituye una categoría independiente y no puede inferirse con precisión a partir de otras variables, estos valores también se conservarán como ausentes.

Conclusión

La mayoría de los valores ausentes se concentran en variables relacionadas con calificaciones de usuarios, críticas especializadas y clasificaciones ESRB. Debido a que estos datos faltantes probablemente reflejan la ausencia real de información y no errores de captura, se decidió mantenerlos como valores nulos para preservar la integridad del conjunto de datos. Únicamente los registros con valores ausentes en name y genre serán eliminados debido a su escasa cantidad y a la importancia de estas variables para identificar correctamente cada videojuego.

````python
#Eliminar valores ausentes 
games = games.dropna(subset=['name', 'genre'])
#verificar valores ausentes
games.isna().sum()
````

````python
#Crear la columna ventas totales
games['total_sales'] = (
    games['na_sales']
    + games['eu_sales']
    + games['jp_sales']
    + games['other_sales']
)
````

## Analisis de datos

````python
#Analizar cantidad de lanzamientos por año 
games_per_year = games.groupby('year_of_release')['name'].count().reset_index()

plt.figure(figsize=(12,6))
plt.bar(games_per_year['year_of_release'], games_per_year['name'])

plt.title('Cantidad de juegos lanzados por año')
plt.xlabel('Año')
plt.ylabel('Número de juegos')
plt.show()
````

Analizar la cantidad de lanzamientos por año permite identificar períodos de crecimiento y madurez de la industria. Vemos que en los primeros años se presentan muy pocos registros, es probable que no sean representativos del mercado actual y aporten poco valor para pronosticar tendencias futuras.

Algunos analisis pueden ser:

Crecimiento inicial (1980-1994)

Durante los primeros años se observa una cantidad relativamente baja de lanzamientos, generalmente inferior a 100 juegos por año. Esto refleja una etapa temprana de la industria, donde el mercado de videojuegos era más pequeño y las plataformas disponibles eran limitadas.

Expansión acelerada (1995-2002)

A partir de mediados de los años noventa se aprecia un crecimiento sostenido en la cantidad de juegos publicados. Este incremento coincide con la popularización de consolas como PlayStation y Nintendo 64, así como con el crecimiento del mercado de videojuegos a nivel mundial.

Período de máximo crecimiento (2002-2011)

Entre 2002 y 2011 la industria alcanza su mayor expansión. La cantidad de juegos lanzados supera los 800 títulos anuales y alcanza su punto máximo alrededor de 2008 y 2009, con más de 1.400 lanzamientos por año.

Este comportamiento puede asociarse a:

La consolidación de plataformas como PlayStation 2, Xbox 360, Wii, Nintendo DS y PlayStation 3.
La expansión global de la industria del entretenimiento digital.
La reducción de barreras tecnológicas para el desarrollo de videojuegos.
Descenso posterior (2012-2016)

Después de 2011 se observa una disminución significativa en el número de lanzamientos. Aunque el mercado de videojuegos continuó creciendo en ingresos, el número de títulos publicados disminuyó.

Algunas posibles razones son:

Transición hacia nuevas generaciones de consolas.
Mayores costos y tiempos de desarrollo.
Concentración de recursos en menos títulos pero de mayor presupuesto.
Auge de modelos de negocio basados en actualizaciones y contenido descargable, reduciendo la necesidad de lanzar tantos juegos nuevos.

````python
#Plataformas con mayores ventas historicas
platform_sales = games.groupby('platform')['total_sales'].sum().sort_values(ascending=False)

print(platform_sales.head(10))
````

````python
#Grafico de mayores ventas historicas
platform_sales.head(10).plot(kind='bar', figsize=(10,6))

plt.title('Top 10 plataformas por ventas totales')
plt.ylabel('Ventas globales (millones)')
plt.show()
````

Este análisis permite identificar las plataformas más exitosas históricamente y sirve como punto de partida para estudiar su ciclo de vida. Ahora realizaremos una Evolución anual de las principales plataformas

````python

#Top de las 6 plataformas mas importantes con mayores ventas historicas
top_platforms = platform_sales.head(6).index

#Gráficos de distribución de las ventas historicas para las 6 plataformas tops por año
platform_year_sales = games[games['platform'].isin(top_platforms)] .pivot_table(
        index='year_of_release',
        columns='platform',
        values='total_sales',
        aggfunc='sum'
    )

platform_year_sales.plot(figsize=(14,7))

plt.title('Evolución anual de ventas por plataforma')
plt.ylabel('Ventas globales')
plt.show()
````

## Pruebas de hipotesis

— Las calificaciones promedio de los usuarios para las plataformas Xbox One y PC son las mismas.

— Las calificaciones promedio de los usuarios para los géneros de Acción y Deportes son diferentes.

Para esta sección del proyecto se realizarán las dos pruebas de hipótesis utilizando una prueba t para muestras independientes (Welch's t-test), ya que se compararán las medias de dos grupos distintos y no es necesario asumir varianzas iguales, con un nivel designificancia (alpha) del 5%

Hipótesis 1: Las calificaciones promedio de los usuarios para Xbox One y PC son las mismas
Formulación de hipótesis

Hipótesis nula (H₀):

La calificación promedio de los usuarios para los juegos de Xbox One es igual a la de los juegos de PC.

Hipótesis alternativa (H₁):

La calificación promedio de los usuarios para los juegos de Xbox One es diferente a la de los juegos de PC.

Nivel de significancia

Este es el nivel de significancia más utilizado en análisis estadísticos y representa un riesgo del 5% de rechazar incorrectamente una hipótesis nula verdadera.

````python
# Filtrar datos válidos
xone_scores = games[(games['platform'] == 'XOne') &(games['user_score'].notna())]['user_score']

pc_scores = games[(games['platform'] == 'PC') & (games['user_score'].notna())]['user_score']

# Prueba t de Welch
results = stats.ttest_ind(xone_scores, pc_scores, equal_var=False)

print("Estadístico t:", results.statistic)
print("p-value:", results.pvalue)

alpha = 0.05

if results.pvalue < alpha:
    print("Rechazamos la hipótesis nula")
else:
    print("No podemos rechazar la hipótesis nula")
````

Análisis

El valor p obtenido (0.00000494) es considerablemente menor que el nivel de significancia establecido (0.05). Esto indica que la probabilidad de observar una diferencia tan grande entre las medias si realmente fueran iguales es extremadamente baja.

Por lo tanto, existe evidencia estadística suficiente para rechazar la hipótesis nula.

Además, el estadístico t negativo sugiere que, en promedio, una de las plataformas presenta calificaciones de usuarios inferiores a la otra. Para determinar exactamente cuál tiene la media más alta sería necesario comparar directamente los promedios de user_score para Xbox One y PC.

Conclusión

Los resultados de la prueba indican que las calificaciones promedio de los usuarios para Xbox One y PC no son iguales. La diferencia observada es estadísticamente significativa, por lo que se rechaza la hipótesis nula y se acepta que existen diferencias en la valoración que los usuarios otorgan a los juegos de ambas plataformas

Hipótesis 2: Las calificaciones promedio de los usuarios para Acción y Deportes son diferentes

Formulación de hipótesis

Hipótesis nula (H₀):

Las calificaciones promedio de los usuarios para los géneros Acción y Deportes son iguales.

Hipótesis alternativa (H₁):

Las calificaciones promedio de los usuarios para los géneros Acción y Deportes son diferentes.

````python
action_scores = games[(games['genre'] == 'Action') & (games['user_score'].notna())]['user_score']

sports_scores = games[(games['genre'] == 'Sports') & (games['user_score'].notna())]['user_score']

results = stats.ttest_ind( action_scores, sports_scores, equal_var=False)

print("Estadístico t:", results.statistic)
print("p-value:", results.pvalue)

alpha = 0.05

if results.pvalue < alpha:
    print("Rechazamos la hipótesis nula")
else:
    print("No podemos rechazar la hipótesis nula")
````

Análisis

El valor p obtenido (0.1148) es mayor que el nivel de significancia establecido (0.05). Esto significa que la diferencia observada entre las calificaciones promedio de los géneros Acción y Deportes podría deberse al azar y no necesariamente a una diferencia real en las poblaciones.

Por esta razón, no existe evidencia estadística suficiente para rechazar la hipótesis nula.

Aunque las medias muestrales pueden presentar pequeñas diferencias, estas no son lo suficientemente grandes como para concluir que los usuarios califican de manera distinta ambos géneros.

Conclusión

Los resultados de la prueba indican que no se encontraron diferencias estadísticamente significativas entre las calificaciones promedio de los usuarios para los géneros Acción y Deportes.

Por lo tanto, no podemos rechazar la hipótesis nula y concluimos que, con la información disponible, las valoraciones de los usuarios para ambos géneros son estadísticamente similares.
